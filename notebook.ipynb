```python
import pandas as pd
```

```python
df = pd.read_csv("dataset.csv")
```

```python
df = df.dropna().drop_duplicates()
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder

encoders = {}

for col in df.columns:
    le = LabelEncoder()
    if type(df[col][0]) == str:
        df[col] = le.fit_transform(df[col])
        encoders[col] = le
```

```python
df
```

```python
encoders
```

```python
X = df.drop(columns=["Exam_Score", "Gender"], axis=1) # i decided to remove gender variable as it is not a very important variable for predicting
y = df["Exam_Score"]
```

```python
X
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2)
```

```python
# i tested various ml algorithms but linear regression performed the best

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_squared_error

y_pred = model.predict(X_test)
mean_squared_error(y_test, y_pred)
```

```python
# accuracy doesn't matter a lot with this model since the student scores could be diverse and what matters more is precision or how close was the predicted value to the student's score, so mean squared error matters more here

model.score(X_test, y_test)
```
